```python
import numpy as np
import pandas as pd
import os

def generate_synthetic_futures_data(
    n_minutes=2000,
    start_price=4500.0,
    tick_size=0.25,
    seed=42
):
    np.random.seed(seed)

    prices = [start_price]
    volumes = []
    bid_volumes = []
    ask_volumes = []
    spreads = []

    volatility = 0.0005

    for i in range(n_minutes):
        # Volatility regime shift every 200 minutes
        if i % 200 == 0 and i > 0:
            volatility = np.random.uniform(0.0003, 0.0012)

        ret = np.random.normal(0, volatility)
        prices.append(prices[-1] * (1 + ret))

        spread_ticks = np.random.choice([1, 2, 3], p=[0.6, 0.3, 0.1])
        spread = spread_ticks * tick_size
        spreads.append(spread)

        base_volume = np.random.randint(800, 2000)
        imbalance = np.random.uniform(-0.5, 0.5)

        bid_volumes.append(max(int(base_volume * (1 + imbalance)), 1))
        ask_volumes.append(max(int(base_volume * (1 - imbalance)), 1))
        volumes.append(base_volume)

    prices = prices[1:]  # align lengths

    df = pd.DataFrame({
        "close": prices,
        "bid": np.array(prices) - np.array(spreads) / 2,
        "ask": np.array(prices) + np.array(spreads) / 2,
        "bid_volume": bid_volumes,
        "ask_volume": ask_volumes,
        "volume": volumes
    })

    return df


def generate_and_save_datasets(
    output_dir="synthetic_data",
    n_datasets=20,
    n_minutes=3000,
    start_seed=100
):
    os.makedirs(output_dir, exist_ok=True)

    for i in range(n_datasets):
        seed = start_seed + i
        df = generate_synthetic_futures_data(
            n_minutes=n_minutes,
            seed=seed
        )

        file_path = os.path.join(
            output_dir,
            f"synthetic_futures_seed_{seed}.csv"
        )

        df.to_csv(file_path, index=False)
        print(f"Saved {file_path}")


# -------------------------
# RUN DATA GENERATION
# -------------------------
generate_and_save_datasets(
    output_dir="synthetic_data",
    n_datasets=20,
    n_minutes=3000,
    start_seed=100
)
```
