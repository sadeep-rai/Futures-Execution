```python
from collections import deque

# Start with a queue of tasks
tasks = deque(['task1', 'task2', 'task3'])

# Add a new task to the end (fast)
tasks.append('task4')

# Handle the oldest task (fast)
oldest = tasks.popleft()

print(f"Handled: {oldest}")
print(f"Remaining queue: {tasks}")
```

```python
print(oldest)
```

```python
class MyEnv:
    def __init__(self):
        self.current_step = 0  # instance variable

    def next(self):
        self.current_step += 1

# Create two different environment instances
env1 = MyEnv()
env2 = MyEnv()

env1.next()
env1.next()
env1.next()
env1.next()
env2.next()

print(env1.current_step)  # 2
print(env2.current_step)  # 0
```

```python
print(env1.next())
```

```python
# !pip install gym

import pandas as pd
import gymnasium as gym
from gym import spaces
import numpy as np

# 1. Prepare your data (simplified example)
data = pd.DataFrame({
    'open': [100, 101, 102],
    'high': [102, 103, 104],
    # ... other OHLCV columns
})

class MyEnv(gym.Env):
    def __init__(self, data, order_size, time_horizon, adv):
        
        super().__init__()
            
            
        self.data = data  # OHLCV (Open High Low Close Volume) + order book data
        self.order_size = order_size
        self.time_horizon = time_horizon  # in minutes
        self.current_step = 0 ## setting up the time-counter to the initial historical point
        self.adv = adv
    
        # Action space: we require 3 distinctive actions for now: 0=Passive, 1=Moderate, 2=Aggressive
        self.action_space = spaces.Discrete(3)
    
            # State space: we have 8 normalized features
        self.observation_space = spaces.Box(
        low=0.0, high=1.0,
        shape=(8,),
        dtype=np.float32)
    
        self.reset()
            
# 2. Create environment - only provide the 4 parameters from __init__
env = MyEnv(
    data=data,              # Required
    order_size=500,         # Optional (default: 1000)
    time_horizon=30,        # Optional (default: 60)
    adv=2_500_000           # Optional (default: 1,000,000)
)

# 3. Now you can access ALL attributes:
print(env.action_space)      # ← Was set automatically
print(env.observation_space) # ← Was set automatically
# print(env.data)              # ← Your provided data
print(env.order_size)        # ← Your provided value or default
```
